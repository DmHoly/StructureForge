```python
from structureforge import (
    Deposition,
    Etch,
    FacetedGrowth,
    Geometry,
    Length,
    Lithography,
    default_library,
    default_recipes,
    save_svg,
    simulate,
)
```

```python
materials = default_library()
recipes = default_recipes()
geometry = Geometry.substrate(
    "GaN",
    domain_width_nm=2000,
    thickness_nm=50,
)

steps = [
    Deposition(name="Dépôt", material="Si3N4", recipe="ALD Conformal", thickness=Length(value=100, unit="nm")),
    Lithography(name="Lithographie", resist_material="Photoresist", thickness=Length(value=50, unit="nm"), openings=[(150, 250), (550, 650), (950, 1050), (1350, 1450), (1750, 1850)]),
    Etch(name="Gravure", recipe="Anisotropic RIE", depth=Length(value=100, unit="nm")),
    FacetedGrowth(name="Croissance facettée", material="GaN", thickness=Length(value=1000, unit="nm"), rate_c=0.8, rate_m=0, rate_sp=0.6, semi_polar_angle_deg=40, seed_materials=["GaN"]),
    FacetedGrowth(name="Croissance facettée", material="In0.35Ga0.65N", thickness=Length(value=50, unit="nm"), rate_c=0.2, rate_m=0, rate_sp=0.95, semi_polar_angle_deg=40, seed_materials=["GaN"]),
]

frames = simulate(geometry, steps, materials, recipes)
material_colors = {m.name: m.color for m in materials}
```
